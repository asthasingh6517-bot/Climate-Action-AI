CONVERTING DATA 

```python
df["Date"] = pd.to_datetime(df["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
```

CREATING TEMPERATURE STRESS

```python
df["Temperature_Stress"] = (
    df["Temperature_Max"] - df["Temperature_Avg"]
)
```

```python
df["Rainfall_Level"] = pd.cut(
    df["Rainfall"],
    bins=[-1, 0, 10, 50, float("inf")],
    labels=["No Rain", "Low", "Moderate", "Heavy"]
)
```

```python
aqi_mapping = {
    "Satisfactory": 1,
    "Moderate": 2,
    "Poor": 3,
    "Very Poor": 4,
    "Severe": 5
}

df["AQI_Severity"] = df["AQI_Category"].map(aqi_mapping)
```

```python
df[[
    "City",
    "Temperature_Avg",
    "Temperature_Stress",
    "Rainfall",
    "Rainfall_Level",
    "AQI",
    "AQI_Category",
    "AQI_Severity"
]].head(10)
```

HEAT RISK

```python
df["Heat_Risk"] = pd.cut(
    df["Temperature_Avg"],
    bins=[-float("inf"), 25, 30, 35, float("inf")],
    labels=[1, 2, 3, 4]
).astype(int)
```

AIR QUALITY RISK

```python
df["Air_Risk"] = pd.cut(
    df["AQI"],
    bins=[-float("inf"), 50, 100, 200, 300, float("inf")],
    labels=[1, 2, 3, 4, 5]
).astype(int)
```

RAINFALL RISK

```python
df["Water_Risk"] = pd.cut(
    df["Rainfall"],
    bins=[-1, 0, 5, 20, float("inf")],
    labels=[4, 3, 2, 1]
).astype(int)
```

OVERALL CLIMATE SCORE


```python
df["Climate_Risk_Score"] = (
    df["Heat_Risk"] * 0.35 +
    df["Air_Risk"] * 0.40 +
    df["Water_Risk"] * 0.25
)
```

```python
df["Climate_Risk"] = pd.cut(
    df["Climate_Risk_Score"],
    bins=[0, 1.75, 2.75, 3.75, float("inf")],
    labels=["Low", "Moderate", "High", "Severe"]
)
```

```python
print(df["Climate_Risk"].value_counts())
```

Showing percentage 

```python
risk_distribution = df["Climate_Risk"].value_counts(normalize=True) * 100

print(risk_distribution.round(2))
```

Visualize the risk distribution

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Climate_Risk",
    order=["Low", "Moderate", "High", "Severe"]
)

plt.title("Distribution of Climate Risk Levels")
plt.xlabel("Climate Risk Level")
plt.ylabel("Number of Records")
plt.show()
```

Checking risk by city

```python
city_risk = pd.crosstab(
    df["City"],
    df["Climate_Risk"]
)

city_risk
```

Visualizing city by risk

```python
city_risk.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Climate Risk Distribution Across Cities")
plt.xlabel("City")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.legend(title="Climate Risk")
plt.tight_layout()
plt.show()
```
